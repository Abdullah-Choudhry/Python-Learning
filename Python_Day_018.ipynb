**Short** **Hand** **Way**

```python
marks = 75
result = "Pass" if marks >= 50 else "Fail"

print(result)
```

```python
age = 16

print("Adult") if age >= 18 else print("Minor")
```

```python
score = 65

print("Pass - Good" if score >= 70 else ("Pass - Average" if score >= 50 else "Fail"))
```

**Enumerate** **Function**

```python
marks = [33, 43, 84, 28, 82, 58]

for index, mark in enumerate(marks):
    print(mark)
    if index==3:
        print("Gooooooooooood")
    index += 1
```

```python
clients = ["Ali", "Ahmed", "Sara", "Zainab"]

for index, client_name in enumerate(clients, start=1):
    print(f"Client #{index}: {client_name} ")
```
