**Inheritance**

```python
class Animal:
    def eat(self):
        print("This animal is eating food")

    def details(self,name):
      print(f'My name is {name}')
class Dog(Animal):
    def bark(self):
        print("The dog is barking Woof Woof!")

my_dog = Dog()
my_dog.bark()
my_dog.eat()

my_dog2 = Dog()
my_dog2.details('Cat')
my_dog2.bark()
```

```python
class Electronics:
    def power_on(self):
        print("Device is turning ON")

class Smartphone(Electronics):
    def take_photo(self):
        print("Taking a high-resolution photo")

my_phone = Smartphone()

my_phone.take_photo()
my_phone.power_on()
```
