**Tell and Seek**

```python
with open('demo.txt', 'w') as f:
    f.write("Hello World")

with open('demo.txt', 'r') as f:
    print(f.tell())


    content = f.read(5)
    print("Read first 5 :", content)

    print("Position after reading 5 chars:", f.tell())

    f.seek(0)
    print("Position after seek(0):", f.tell())

    print("Read again from start:", f.read())
```

```python
with open('sample.txt', 'w') as f:
    f.write("Python Programming")

with open('sample.txt', 'r') as f:
    f.seek(3)
    print("Current position:", f.tell())

    content = f.read()
    print("Content after seek(3):", content)
```

```python
with open('simple_truncate.txt', 'w') as f:
    f.write("Python Programming")
    f.truncate(6)

with open('simple_truncate.txt', 'r') as f:
    print(f.read())
```
