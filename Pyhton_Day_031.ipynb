**Class Method**

```python
class School:
    school_name = "Superior Academy"

    @classmethod
    def change_school_name(cls, new_name):
        cls.school_name = new_name

print(School.school_name)

School.change_school_name("Al-Falah Academy")

print(School.school_name)
```

**As constructor**

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    @classmethod
    def from_string(cls, data_str):
        name, age = data_str.split("-")
        return cls(name, int(age))

s1 = Student("Ali", 20)
s2 = Student.from_string("Ahmed-22")
print(s1.name)
print(s2.name)
print(s2.age)
```
