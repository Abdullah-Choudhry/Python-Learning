**Dictionary** and **Methods**

```python

student = {
    11 : "Abdullah",
    "age": 20,
    "course": "Artificial Intelligence"
}

print(student[11])
print(student["course"])
print(student["age"])
print(student.get("age"))
print(student.get("age2"))
print(student.keys())
print(student.values())
for key in student.keys():
  print(f"value is {student[key]}")


print(student.items())
```

```python
ep1 = {89: 19, 33: 43, 55: 22}

ep2 = {11: 99, 76: 54}

ep1.update(ep2)
print(ep1)
ep2.clear()
print(ep2)
```

```python
dic = {
      "school": "no",
       "age": 5,
       22: "abdullah"
       }

dic.popitem()
print(dic)
dic.pop("school")
print(dic)
```

```python

student = {
    "name": "Abdullah",
    "age": 20,
    "city": "Lahore"
}

print(student["name"])
print(student["age"])


student["grade"] = "A"
print(student)
del student["age"]
print(student)
```
