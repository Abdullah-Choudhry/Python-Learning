**Decorators**

```python
def decorator(func):
    def wrapper(*args,**kwargs):
        print("Good Morning!")
        func(*args,**kwargs)
        print("Thanks for using this function.")
    return wrapper

@decorator

def welcome():
    print("Welcome to the Python Object-Oriented Programming module.")

@decorator
def add(a,b):
  print(a+b)


# 3. Calling the Decorated Function
welcome()
add(2,3)
```

**Getters and Setters**

```python
class Student:
    def __init__(self, name, age):
        self.__age = age

    def get_age(self):
        return self.__age

    def set_age(self, new_age):
        if new_age > 0:
            self.__age = new_age
        else:
            print("Invalid age entered!")


s1 = Student("Abdullah", 20)

print("Current Age:", s1.get_age())

s1.set_age(22)
print("Updated Age:", s1.get_age())
```
