**for** **loops**

```python
name = "abdullah"
for i in name:
  print(i)
  if(i=="u"):
    print("special one")
colors = ["red", "green", "blue", "black",]
print()
for color in colors:
  print(color)
  for b in color:
    print(b)

print()
for k in range(5):
  print(k + 1) #1 to 5
  print()
for p in range(1, 10, 2):
  print(p)
```

While Loop

```python
i = int(input("Enter the number:"))
while(i<=40):
  i = int(input("Enter the number:"))
  print(i)
print("Done with the loop")
```

```python
u = 5
while (u>0):
  print(u)
  u = u-1
else:
  print("Else case")
```
