**Classes and Objects**

```python
class person:
  name = 'Rohan'
  age = 18
  profession = 'Engineer'

a = person()
a.age = 19
a.profession = 'Developer'
print(a.name, a.age, a.profession)
```

```python
class Person:
    name = 'Rohan'
    age = 18
    profession = 'Engineer'
    def show_details(self):
        print(f"Name: {self.name}, Age: {self.age}, Profession: {self.profession}")
        print(f'{self.name }, is a {self.profession}')

a = Person()

a.age = 19
a.profession = 'Developer'

a.show_details()
```


**Constructor**

```python
class Person:
    # __init__ automatic chalta hai jab object banta hai
    def __init__(self, name, age, profession):
        self.name = name
        self.age = age
        self.profession = profession

    def show_details(self):
        print(f"Name: {self.name}, Age: {self.age}, Profession: {self.profession}")

# Ab ek hi line mein naya student/person ban gaya!
person1 = Person("Rohan", 19, "Developer")
person2 = Person("Abdullah", 20, "AI Engineer")

person1.show_details()
person2.show_details()
```
