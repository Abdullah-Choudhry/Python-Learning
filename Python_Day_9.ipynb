**Lists**

```python
l = [3, 5, 7, 'Hello', 9 ,56, 78]
print(l)
marks = [2, 8, 9]
print(type(marks))
print(marks[1])
print(marks[0])
print(marks[-1])
print(marks[len(marks) -1])
print(marks[3-1])
if 8 in marks:
  print("Yes")
else:
  print("No")
if 5 in l:
  print("Yes")
else:
  print("No")
print(l[1:6])
print(l[:])
print(l[:2])
print(l[:6:2])
```

**Lists**  **Comprehension**

```python
lst = [i for i in range(4)]
print(lst)
lst = [i*i for i in range(10) if (i%2)]
print(lst)
```

**Lists** **Method**

```python
l = [77, 8, 1, 2, 3, 4]
print(l)
l.append(7)
print(l)
l.sort
print(l)
l.sort(reverse=True)
print(l)
l.reverse()
print(l)
print(l.index(8))
print(l.count(2))
n = l.copy()
n[1] = 0
print(l)
l.insert(2, 88)
print(l)
```
