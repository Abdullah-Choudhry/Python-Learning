**Super** **Keyword**

```python
class Person:
    def __init__(self, name):
        self.name = name

class Student(Person):
    def __init__(self, name, roll_no):
        super().__init__(name)
        self.roll_no = roll_no

s1 = Student("Abdullah", "SU92-BSAIM-S26-030")

print(s1.name)
print(s1.roll_no)
```

```python
class Account:
    def show_info(self):
        print("Account Holder: Valid User")

class SavingsAccount(Account):
    def show_info(self):
        super().show_info()
        print("Account Type: Savings Account")

acc = SavingsAccount()
acc.show_info()
```
