**Lamda** **Function**

```python
add_ten = lambda x: x + 10

print(add_ten(5))
```

```python
multiply = lambda a, b: a * b

print(multiply(4, 5))
```

```python
square = lambda x: x ** 2

print(square(4))

cube = lambda x: x ** 3

print(cube(3))
```

**Map**

```python
numbers = [1, 2, 3, 4, 5]

new_numbers = list(map(lambda x: x * 2, numbers))

print(new_numbers)
```

**Filter**

```python
numbers = [1, 2, 3, 4, 5, 6]

even = list(filter(lambda x: x % 2 == 0, numbers))

print(even)
```

 **Reduce**

```python
from functools import reduce

a = [1, 2, 3, 4]

b = reduce(lambda x, y: x * y, a)

print(b)
```
