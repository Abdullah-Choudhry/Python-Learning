**Sets** **Methosds**

```python
a = {1, 2, 4, 5, 9}
b = {0, 3, 1}
print(a.union(b))
a.update(b) #adding value of b in a
print(a)
cities1 = {"Tokyo", "Lahore", "Paris"}
cities2 = {"Tokyo", "Madrid"}
print(cities1.difference(cities2))
print(cities1.intersection(cities2))
cities3 = cities1.symmetric_difference(cities2)
print(cities3)
cities1.intersection_update(cities2)
print(cities1)

```

```python
# Example 1: when no common in stes (Disjoint sets)
team_a = {"Ali", "Ahmed"}
team_b = {"Sara", "Zain"}

print(team_a.isdisjoint(team_b))


# Example 2:
group_1 = {"apple", "banana", "cherry"}
group_2 = {"mango", "banana", "orange"}

print(group_1.isdisjoint(group_2))
```

**issubet** , **issuperset**

```python
x = {"apple", "banana"}
y = {"apple", "banana", "cherry", "mango"}
print(x.issubset(y))



y = {"apple", "banana", "cherry", "mango"}

x = {"apple", "banana"}
print(y.issuperset(x))
```

**Remove** and **Discard**

```python
my_set = {"apple", "banana", "cherry"}
my_set.remove("banana")
my_set.discard("mango")     # discard ignore that element and code runs
print(my_set)

# my_set.remove("mango")  # error because mango is not present in set
```

POP , DEL , CLEAR

```python
my_set = {10, 20, 30, 40}
my_set.pop()                 # one element will remove
print(my_set)



my_set = {10, 20, 30}

my_set.clear()

print(my_set)   # ALL elements cleared

# my_set = {10, 20, 30}

# del my_set                # set will delete
```
