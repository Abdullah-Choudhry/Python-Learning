**is and ==**

```python
list1 = [1, 2, 3]
list2 = [1, 2, 3]

print(list1 == list2)

print(list1 is list2)
```

```python
a = 4
b = 4

print(a == b)
print(a is b)
```

```python
x = [1, 2, 3]
y = x

print(x == y)
print(x is y)
```
