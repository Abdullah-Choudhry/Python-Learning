**Exception** **Handling**

```python
try:
    num = int(input("Enter a number: "))
    result = 10 / num
    print(f"The result is: {result}")

except ZeroDivisionError:
    print("Error: You cannot divide a number by zero (0)!")
except ValueError:
    print("Error: Please enter a valid integer, letters are not allowed!")
```

```python
students_list = ["Ali", "Ahmed", "Sara"]

try:
    # Asking user for an index number
    index = int(input("Enter student index to search (0-2): "))
    print(f"Student found: {students_list[index]}")

except ValueError:
    print("Error: Please enter a valid number, not letters!")
except IndexError:
    print("Error: Index out of range! Please choose a number between 0 and 2.")
```

**Finally** **keyword**

```python
try:
    print("Trying to open and process data...")
    num = int(input("Enter a number: "))
    result = 10 / num
    print(f"Result: {result}")

except ZeroDivisionError:
    print("Error: Cannot divide by zero!")
except ValueError:
    print("Error: Invalid input, please enter a number!")

finally:
    print("Cleanup")
```
