**Static Method**

```python
class Math:
    @staticmethod
    def add(a, b):
        return a + b

result = Math.add(5, 10)
print(result)
```


**Variables**

```python
class Student:
    university_name = "Superior University"   #class variable

    def __init__(self, name, roll_no):
        self.name = name                      #instance variable
        self.roll_no = roll_no

s1 = Student("Abdullah", 30)
s2 = Student("Ali", 45)

print(s1.name)
print(s2.name)

print(s1.university_name)
print(s2.university_name)
```

```python
class Dog:
    # Class Variable
    species = "Canine"
    total_dogs = 0

    def __init__(self, name, breed):
        self.name = name
        self.breed = breed

        Dog.total_dogs += 1

dog1 = Dog("Rex", "German Shepherd")
print(dog1.name, dog1.species, Dog.total_dogs)

dog2 = Dog("Max", "Bulldog")
print(dog2.name, dog2.species, Dog.total_dogs)

dog1.name = "Rocky"
print(dog1.name)
print(dog2.name)

Dog.species = "Wild Canine"
print(dog1.species)
print(dog2.species)
```
