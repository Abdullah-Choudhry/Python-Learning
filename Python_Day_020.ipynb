**OS** **Module**

```python
import os

# Main folder check
if not os.path.exists("data"):
    os.mkdir("data")

# 10 Folders safely create karna
for i in range(0, 10):
    folder_path = f"data/Day{i+1}"
    if not os.path.exists(folder_path):  # Safety check
        os.mkdir(folder_path)

print("10 Folders Done")


if os.path.exists("data/Day1"):
    os.rename("data/Day1", "data/Day_01")
    print("Day1 successfully renamed to Day_01!")
```
