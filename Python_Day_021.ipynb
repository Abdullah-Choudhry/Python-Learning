**Local vs Global Variables**

```python
x = 5      # Global variable
def function():
  x = 4    # Local variable
  print(x)
  print('Hello')

function()

print(x)
```

**Changing Global variable**

```python
x = 5      # Global variable
def function():
  global x   # Value Chnaged
  x = 4    # Local variable
  print(x)
  print('Hello')

function()

print(x)
```
