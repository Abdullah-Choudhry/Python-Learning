```python
age = int(input("Enter number: "))

if age < 18:
    raise ValueError("Error: Age is less than 18, access denied!")

print("The rest of the code executed successfully.")
```

```python
account_balance = 500
withdrawal_amount = 800

if withdrawal_amount > account_balance:
    raise Exception("Error: Insufficient balance in your account!")

print("Transaction successful. Please collect your cash.")
```

```python
user_input = input("Enter a number between 5 and 10 (or type 'quit'): ")

if user_input == 'quit':
    print("Program exited")

else:

    if user_input == 'QUIT' or not user_input.isdigit():
        raise ValueError("Error")

    number = int(user_input)

    if 5 <= number <= 10:
        print(f"Success, Valid number : {number}")
    else:
        raise ValueError(" Number must be between 5 and 10!")
```
